```python
8.
matriz = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

soma = 0

for linha in matriz:
    soma += linha[1]

print(soma)
```

```python
9.
matriz = [
    [12, 5, 8],
    [20, 3, 15],
    [7, 25, 10]
]

maior = matriz[0][0]

for linha in matriz:
    for numero in linha:
        if numero > maior:
            maior = numero

print(maior)
```

```python
10.
notas = [
    [7, 8, 9],
    [6, 5, 8],
    [10, 9, 7]
]

# a) Exibir todas as notas do primeiro aluno
print("a)", notas[0])

# b) Exibir a segunda nota do segundo aluno
print("b)", notas[1][1])

# c) Exibir a terceira nota de cada aluno
print("c)")
for aluno in notas:
    print(aluno[2])

# d) Alterar a primeira nota do terceiro aluno para 8
notas[2][0] = 8
print("d)", notas)
```
