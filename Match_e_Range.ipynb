```python
cor = 1


match cor:
    case 1:
        print("Cor azul")
    case _:
        print("Outra cor")
```

```python
numero = 2

match numero:
  case 1:
    print("numero Um")
  case 2:
    print("numero Dois")
  case 3:
    print("numero três")
```

```python
Bebidas =2

match Bebidas:
  case 1:
    print("Àgua")
  case 2:
   print("Suco")
  case 3:
     print("Refrigerante")
```

```python
dia = "sábado"

match dia:
    case "sábado" | "domingo":
        print("Final de semana")
    case _:
        print("Dia de semana")
```

```python
nota = "A"

match nota:
    case "A":
        print("Excelente")
    case "B":
        print("Bom")
    case "C":
        print("Precisa melhorar")
    case _:
        print("Nota inválida")
```
